```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
)
```

```python
from pydantic import BaseModel, Field

class ResponseStructure(BaseModel):
    name: str = Field(..., description="The name of the person")
    age: int = Field(..., description="The age of the person")
    email: str = Field(..., description="The email address of the person")
structured_output = llm.with_structured_output(
    ResponseStructure,
    method="json_schema",
)
```

```python
text = "John Doe is a software engineer who lives in San Francisco. He is 30 years old and his email address is john.doe@example.com."
```

```python
res = structured_output.invoke(f"please give me a name, age and email address of a person from this text: {text}")
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    director: str = Field(..., description="The director of the movie")
    genre: str = Field(..., description="The genre of the movie")

class MovieList(BaseModel):
    movies: list[Movies] = Field(..., description="A list of movies")

structured_output_movies = llm.with_structured_output(MovieList)
```

```python
result = structured_output_movies.invoke("give me the latest 3 movies of mahesh babu")
result.model_dump()
```
