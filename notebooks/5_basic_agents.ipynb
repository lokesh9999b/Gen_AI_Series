```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers. 
       Args: 
          a: Number One
          b: Number Two
    """
    return a + b

@tool
def mul_numbers(a: int, b: int) -> int:
    """Multiply two numbers. 
       Args: 
          a: Number One
          b: Number Two
    """
    return a * b
@tool
def substarction(a: int, b: int) -> int:
    """Subtract two numbers. 
       Args: 
          a: Number One
          b: Number Two
    """
    return a - b
@tool
def division(a: int, b: int) -> int:
    """Divide two numbers. 
       Args: 
          a: Number One
          b: Number Two
    """
    return a / b
@tool
def reminder(a: int, b:int) -> int:
    """Get reminder of two numbers. 
       Args: 
          a: Number One
          b: Number Two
    """
    return a % b
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b", streaming=True)
```

```python
agent = create_agent(llm, tools=[add_numbers, mul_numbers,substarction, division, reminder], system_prompt="you are a math teacher every time use tools for calculations")
```

```python
response = agent.invoke({"messages":[{"role":"user", "content": "what is division of 100000000 with 5 and muliply the answer with 5"}]})
```
