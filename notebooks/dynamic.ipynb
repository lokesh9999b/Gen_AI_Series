```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGroq(model="openai/gpt-oss-20b", streaming= True)
out = StrOutputParser()
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator to {language}. Just you need to translate the text coming from user to {language}. Do not add any extra text or explanation. Just translate the text."},
    {"role": "user", "content": "{query}"},
])
def transform_case(result: str):
    return result.upper()
# final_prompt = prompts.format_prompt(language="Malayalam", query="write the code to check the bank balance?")
# res = llm.invoke(final_prompt.to_messages())
# print(res.content)


# res = llm.invoke(prompts.invoke({"language": "bengali", "query": "Lokesh"}))
# res.content

chains = prompts | llm | out | transform_case
res = chains.invoke({"language": "Telugu", "query": "write the code to check the bank balance?"})
print(res)
```
