##### Static Prompts


```python
prompts = [
    ("system", "You are a python fullstack developer."), 
    ("user", "write the code for basic login page backend code use mongo db as database")
]
res = llm.invoke(prompts)
print(res.content)
```

```python
#### GROQ MODEL

from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    streaming=True
)

res = llm.stream("Who is prabhas")
for i in res:
    print(i.content, end="")
```
